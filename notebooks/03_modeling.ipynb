```python
# --- Cell 1: Pipeline Pipeline Execution ---
import sys
import os
import pandas as pd

sys.path.append(os.path.abspath(os.path.join('..')))
from src.data_loader import InsuranceDataLoader
from src.modeling import InsurancePredictiveModeler

# Ingest your baseline pipe-delimited data asset
loader = InsuranceDataLoader(file_path="../data/MachineLearningRating_v3.txt")
df = loader.load_and_initialize()

print(f"📊 Dataset Ready. Initializing Machine Learning Sequence...")
```

```python
# --- Cell 2: Train & Evaluate Target Model ---
# Initialize modeling engine
modeler = InsurancePredictiveModeler(df)

# Prepare matrices for Claim Severity (TotalClaims) prediction
print("⚙️ Encoding features and constructing target arrays...")
X, y = modeler.prepare_features_and_targets(target_col='TotalClaims')

# Run the training pipeline
print("🚀 Training Random Forest Regressor (This may take a moment)...")
metrics, top_features = modeler.train_random_forest(X, y)

print("\n🎯 MODEL EVALUATION METRICS COMPLETE:")
print(f"🔹 Mean Absolute Error (MAE): {metrics['MAE']:.2f}")
print(f"🔹 Root Mean Squared Error (RMSE): {metrics['RMSE']:.2f}")
print(f"🔹 R² Variance Score Explained: {metrics['R2_Score'] * 100:.2f}%")

print("\n🔥 TOP 10 STRONGEST RISK COHORT DRIVERS:")
print(top_features.to_string(index=False))
```
